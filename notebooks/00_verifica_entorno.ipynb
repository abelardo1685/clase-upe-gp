```python
import numpy as np
import matplotlib.pyplot as plt
t = np.arange(24)
demanda = 100 + 30*np.sin((t-8)*np.pi/12) + np.random.default_rng(0).normal(0, 3, 24)
plt.plot(t, demanda, "o-", color="#003D79")
plt.xlabel("hora del día"); plt.ylabel("demanda (MW)"); plt.title("Si ves esta gráfica, matplotlib funciona")
plt.show()
```

```python
from sklearn.gaussian_process import GaussianProcessRegressor
gp = GaussianProcessRegressor().fit(t[:, None], demanda)
print("✔ scikit-learn ajustó un proceso gaussiano de prueba")
print()
print("========== ENTORNO LISTO ==========")
```

## Si algo falló

| Mensaje | Qué significa | Solución |

| La celda nunca termina (⏳) | El kernel no arrancó | `pip install ipykernel` y reinicia VS Code |
